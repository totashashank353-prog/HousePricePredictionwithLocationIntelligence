-----------------------------------DATA PREPROCESSING --------------------------

```python
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import OneHotEncoder
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
```

```python
#df.isnull().sum()
#df.duplicated().sum()
df.drop_duplicates()
df
```

```python
X=df.drop("Price", axis=1) # feature selection so drop the target variable
Y=df["Price"]  # only target variable
```

```python
X=X.drop("Id",axis=1) #axis=1 means cols (->) and axis=0 rows (|>)
```

```python
numerical_cols=[
    "Area",
    "Bedrooms",
    "Bathrooms",
    "Floors",
    "YearBuilt"
]
categorical_cols=[
    "Location",
    "Condition",
    "Garage"
]
```

```python
preprocessor=ColumnTransformer(
    transformers=[
        ("num",StandardScaler(),numerical_cols),
        ("cat",OneHotEncoder(drop="first",handle_unknown="ignore"),categorical_cols)
    ]
)
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
X_train_p = preprocessor.fit_transform(X_train)
X_test_p = preprocessor.transform(X_test)
```

OUTLIER DETECTION && TREATMENT

```python
sns.histplot(df["Price"],kde=True)
```

```python
for col in numerical_cols:
    plt.figure(figsize=(6, 4))
    sns.histplot(df[col], kde=True)
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()
```

```python
for col in numerical_cols: # checking the outliers by using the boxplot
  plt.figure(figsize=(6,4))
  sns.boxplot(df[col])
  plt.title("box_plot of col")
  plt.xlabel(col)
  plt.ylabel("Frequency")
  plt.show()
```

```python
# IQR METHOD
q1=df["Price"].quantile(0.25)
q3=df["Price"].quantile(0.75)

IQR=q3-q1

lower_limit=q1-1.5*(IQR)
upper_limit=q3+1.5*(IQR)

print(lower_limit)
print(upper_limit)
print(IQR)
```
